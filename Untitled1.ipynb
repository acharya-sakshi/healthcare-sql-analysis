```python
import pandas as pd
import sqlite3

# Upload the CSV first (run this cell, then choose the file when prompted)
from google.colab import files
uploaded = files.upload()

# Load CSV into a DataFrame
df = pd.read_csv('healthcare_dataset.csv')

# Create a SQLite database and load the data into a table
conn = sqlite3.connect('healthcare.db')
df.to_sql('patients', conn, if_exists='replace', index=False)

# Quick check: preview the table
pd.read_sql_query("SELECT * FROM patients LIMIT 5;", conn)
```

```python
# Question 1: What are the most common medical conditions?
query1 = """
SELECT `Medical Condition`, COUNT(*) as patient_count
FROM patients
GROUP BY `Medical Condition`
ORDER BY patient_count DESC;
"""
pd.read_sql_query(query1, conn)
```

```python
# Question 2: What is the average billing amount by medical condition?
query2 = """
SELECT `Medical Condition`,
       ROUND(AVG(`Billing Amount`), 2) as avg_billing,
       COUNT(*) as patient_count
FROM patients
GROUP BY `Medical Condition`
ORDER BY avg_billing DESC;
"""
pd.read_sql_query(query2, conn)
```

```python
# Question 3: Average length of stay by medical condition
query3 = """
SELECT `Medical Condition`,
       ROUND(AVG(JULIANDAY(`Discharge Date`) - JULIANDAY(`Date of Admission`)), 1) as avg_stay_days,
       COUNT(*) as patient_count
FROM patients
GROUP BY `Medical Condition`
ORDER BY avg_stay_days DESC;
"""
pd.read_sql_query(query3, conn)
```

```python
# Question 4: Most common conditions by age group
query4 = """
SELECT
  CASE
    WHEN Age < 18 THEN 'Under 18'
    WHEN Age BETWEEN 18 AND 35 THEN '18-35'
    WHEN Age BETWEEN 36 AND 55 THEN '36-55'
    WHEN Age BETWEEN 56 AND 75 THEN '56-75'
    ELSE '75+'
  END as age_group,
  `Medical Condition`,
  COUNT(*) as patient_count
FROM patients
GROUP BY age_group, `Medical Condition`
ORDER BY age_group, patient_count DESC;
"""
pd.read_sql_query(query4, conn)
```

```python
# Question 5: Average billing by admission type and insurance provider
query5 = """
SELECT `Admission Type`,
       `Insurance Provider`,
       ROUND(AVG(`Billing Amount`), 2) as avg_billing,
       COUNT(*) as patient_count
FROM patients
GROUP BY `Admission Type`, `Insurance Provider`
ORDER BY avg_billing DESC
LIMIT 10;
"""
pd.read_sql_query(query5, conn)
```
